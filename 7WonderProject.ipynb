# Abstract 
The goal of this project is to implement and compare AI agents for the board game "7 Wonders". We consider factors such as opponents' hidden hands, simultaneous turns, and delayed scoring, which prevent simple tree search or brute force. We propose a MCTS-based agent for which we can experiment with different rollouts, bandit formulas, etc. for tuning. We will then compare this agent with other agents of our design: a random baseline, a greedy heuristic benchmark, and a heuristic agent with weights tuned through simulated annealing. Performance will be measured over a round-robin style tournament with thousands of simulated games using metrics such as win rate, mean final score, and score margin. Since there is no fixed dataset that exists for a task like this, we will utilize a simulator to run many trials of playing the game and generate the data ourselves. In particular, we will use 7Wonder-RL-Lib, an implementation of the game that is built for training RL-based agents. Each observation would be a game state, including players' victory points, the agent's hand, resources, coins, trading actions, visible board, and action taken.

# Background

Games are no stranger to being the subject of many tests for AI and learning methods. Landmark agents have been created to match human players in many turn-based strategy games, perhaps the most popular being AlphaGo for Go and, by extension, AlphaZero for Chess and Shogi[<sup>[1]</sup>](#chessnote)<a name="chess"></a>. These landmark models use a combination of Monte Carlo tree search (MCTS) and self-play reinforcement learning to learn optimal strategies. In general, this is reflective of the agentic board game literature in general; many existing papers focus on the combination of MCTS and reinforcement learning in their methods. A recent survey of modern tabletop games by Csereoka and Micea found this to be the case, with MCTS being the most common first approach[<sup>[2]</sup>](#surveynote)<a name="survey"></a>. However, their survey also brings to light that the existing literature is concentrated on a small set of classical games, and that literature for newer board games is sparse. In particular, competitive, multiplayer games with hidden information remain comparatively under-studied. Thus, we discuss applications of MCTS to one competitive multiplayer game in particular, called 7 Wonders.

7 Wonders by Antoine Bauza is a card-drafting, civilization-building game for 3 to 7 players played over three stages, called Ages. In each age, all players are dealt a hand of 7 random cards from that age. Each round, all players simultaneously choose a card to build, use toward a wonder stage, or discard for coins, with the remaining cards bening passed to their neighbor. Players may also trade resources with their direct neighbors, and at the end of each age, a military comparison is run with their direct neighbors. The winner is the player with the most Victory Points after the third age. For an agent, this poses numerous challenges, including dealing with simultaneous moves, partial observability, a non-zero-sum multiplayer setting, and a dependence on other players for trading.

Prior work on 7 Wonders is limited, but exists. Robilliard et al. applied MCTS and found it outperformed standard rule-based agents[<sup>[3]</sup>](#robilliardnote)<a name="robilliard"></a> in a three-player setting, including trading, handling hidden hands with determinization and simultaneous moves with a per-player level tree. Against two rule-based opponents, their agent won 68% of games with 125 simulations per turn and 93% with 1,000. A follow-up paper by Robilliard Fonlupt also found genetic programming to be a viable learning strategy, competing with MCTS[<sup>[4]</sup>](#fonluptnote)<a name="robilliard"></a>. Minenko built upon the original work by Robilliard, using joint nodes for simultaneous moves instead of separate player trees and considering discards as a potentially valuable move[<sup>[5]</sup>](#minenkonote)<a name="minenko"></a>, achieving similar results. Paolini et al. used the methods from AlphaZero to create ZeusAI, a MCTS and TNN-based agent for the 2-player variant of the game, 7 Wonders Duel[<sup>[6]</sup>](#paolininote)<a name="paolini"></a>. Although a fundamentally different ruleset, there are still some similarities between the two versions of the game and we can use it as inspiration should we decide to experiment with neural networks.

We identify some gaps in the literature that our project will target. Evaluation metrics are a big one, as evaluations in Robilliard and et al. and Minenko focus primaarily on win rates again hand-designed, rule-based opponents in three-player games only. Although winrate is an important metric, other measures such as mean final VP or score margin may also be interesting evaluation metrics. We will also expand environments to include more than three players, as the original game has a variable player count which is an important part of its gameplay and strategy.

# Problem Statement

We want to build an agent that plays the base game of 7 Wonders well, defined by choosing the action that gives it the best final result against its opponent using what is observable on each turn. The problem comes from the amount of choice at each turn. Players pick one of 2-7 cards (depending on the turn), then decide to either play it, use it to build a wonder, or discard it. So, each turn has up to 21 possible actions per player. This happens 6 times per Age, across 3 Ages, so each player makes 18 decisions. This does not account for trading dynamics or randomness of the deal, which creates an environment way too complex to write down any "good" rules. Ideally, we'd want an agent to learn a good strategy through experience. instead.

Formally, the game is a stochastic game with hidden information and simultaneous moves. An agent follows a policy that maps what it observes (its hand, its own board, its resources, its action space, and its opponents’ visible boards) to one of its legal actions. Thus, the problem is to find a policy that maximizes the expected score margin, defined as the difference between the agent's final points and the final points of the next closest player. For an agent that is winning, the score margin is positive to represent the difference from the person in second place; for an agent that is losing, the score margin is negative.

We will measure success using each agent’s win rate and average score margin over many simulated games with seats/turn orders rotated. We will use our random agent to establish a baseline: with N players at the table, it should have a winrate of rougly $\frac{1}{N}$. Our simulated annealing agent should perform better than that, and then our Monte Carlo search agent should perform even better

The problem is replicable because every game is a new deal of the same decision problem. We can seed and log each game that is played so that the results can be reproduced exactly.

# Data

There's no existing dataset for this, so all of our data will be generated through simulated play in 7Wonder-RL-Lib<a name="rllib"></a>[<sup>[7]</sup>](#rllibnote), an open-source Python environment for 7 Wonders that we'll fork and add our agents to. Our fork, experiment code, and generated data will all be in our GitHub repo.

**Observations:** Each observation is one decision by one agent. It encompasses the game state (hand, resources, coins, actions, opponents' visible boards), the action it took, and that player's final score once the game ended. We may also choose to incorporate information such as cards in opponents' hands that are known after one or more turns. We will keep a smaller summary table with one row per player per game, which is what the evaluation metrics are computed from.

**Size:** Each player makes 18 decisions per game, calculated from 6 turns in each of 3 ages. (For the purposes of our project, we will assume board selection is randomized, rather than player-chosen). Thus, one game produces anywhere from 54 to 126 observations depending on player count. A few thousand games gives us hundreds of thousands of observations and over 10,000 player-game rows, and we can always simulate more if we need. Each observation has a few dozen variables plus the card vectors described below.

**Variables:** We define the following variables for our dataset:
- **Cards** are integer IDs from our card file, so a hand or a set of built cards is a multi-hot vector over all the distinct cards.
- **Boards** are vectors of cards, along with other integer values to store information like the money that player has (which is public information), which board the player has, and which stage of the wonder has been completed.
- **Resources** are a vector of counts for each of the 7 resource types.
- **Coins, shields, and wonder stages** are integers, which may just be encoded in the board state.
- **Actions** are combinations of a card ID plus a type (build, wonder stage, or discard). 
- **Player rows** store the seed used for randomness, player count, Age, turn, seat, wonder board, and agent type. 
- **Outcome variables** include final score (broken down by category), whether the player won, and score margin.

**Special handling:** There's no cleaning in the usual sense, but the data is only as good as the engine we're using. So, we play to validate the rules with unit tests before generating stuff for real. Observations only include what the agent could actually see, Monte Carlo rollouts aren't logged, and tuning and evaluation use separate seeds to prevent leakage. Since turns from the same game are correlated, we'll treat games rather than turns as the unit when computing confidence intervals.

# Proposed Solution

As mentioned above, we will use 7Wonder-RL-Lib, an open-source Python engine for 7 Wonders to generate all of our data. Other code implementations of the game exist, a popular one being a C++ engine from hadim et al.<a name="jardim"></a>[<sup>[8]</sup>](#jardimnote). We choose 7Wonder-RL-Lib for its greater ease of implementing agents and, since it's in Python, the Monte Carlo agent can copy game states for its simulations, although we may need to make some modifications to the engine for that. Each turn, every agent gets an observation (its hand, resources, coins, opponents' visible boards, etc.) and a list of legal actions. Everyone independently chooses at once, and the environment resolves the turn and passes hands. As we will mention in our Ethics section, the cheapest way to pay for a card is picked automatically, so an action is just a card plus whether to build it, use it for a wonder stage, or sell it. Every game is seeded so it can be replayed, and each turn's observation and action is logged along with that player's final score.

**Random baseline and greedy agent:** The random baseline is the library's built-in random player, which picks uniformly from its legal actions. The greedy heuristic agent picks the action that maximizes a hand-weighted evaluation $V_w(s) = w^\top \phi(s)$, where $\phi(s)$ holds features like current points, resources produced, coins, military strength versus each neighbor, and wonder stages built.

**Tuned heuristic agent:** This agent uses the same features a similar decision rule, but its weights are instead tuned with simulated annealing, as defined by Kirkpatrick et al.<a name="kirkpatrick"></a>[<sup>[9]</sup>](#kirkpatricknote). In other words, the tuning is the only difference from the benchmark. Each step adds Gaussian noise to the weights and scores the candidate $w'$ by its mean score margin $f(w')$ over $K$ games against random and greedy opponents, always keeping better candidates and keeping worse ones with probability $e^{(f(w') - f(w))/T}$ as the temperature $T$ cools. We'll experiment with the starting temperature, cooling rate, step size, and $K$.

**Monte Carlo search agent:** For each legal action, this agent runs simulations that fill its opponents' hands with random unseen cards (known as determinization<a name="cowling"></a>[<sup>[10]</sup>](#cowlingnote)) and then play the game to the end with a fast rollout policy for every player, picking the action with the best average result. This deals with hidden hands, simultaneous selection, and delayed scoring directly. We plan to start with regular Monte Carlo with random rollouts, which is like some of the previous literature, then move to full MCTS and optionally other rollout functions or bandit strategies<a name="browne"></a>[<sup>[11]</sup>](#brownenote) as time allows, tuning the number of simulations per move along the way.

**Evaluation:** Monte Carlo search doesn't need a hand-written evaluation function and has already beaten rule-based players in this game per prior literature<a name="robilliard"></a>[<sup>[3]</sup>](#robilliardnote). Robilliard and Fonlupt also found that players evolved with genetic programming could compete with MCTS in 7 Wonders<a name="fonlupt"></a>[<sup>[4]</sup>](#fonluptnote), which suggests a well-tuned heuristic could come close at a much lower cost. Thus, the round-robin tournament with many simulations will be used to collect data which we then calculate our evluation metrics from while comparing them to the random baseline and greedy heuristic. Before the tournament, we'll unit test the library's scoring and payment rules against hand-worked examples and check that random agents in every seat each win about $1/N$ of games.

# Evaluation Metrics

As a recap of our evaluation process, we'll compare the four agents in a round-robin tournament of thousands of simulated games, with the greedy heuristic as our benchmark and results reported separately for each player count from 3 to 7. In head-to-head games one copy of an agent plays against $N-1$ copies of another, and in mixed games all four agents share a 4-player table. Seats will be rotated between agents with random wonder boards and matchups will use the same seeds. We identify the following metrics which we will use to evaluate the effectiveness of our agents:

**Win rate**: the fraction of games agent $a$ finishes first. We express this as $\text{WR}_a = \frac{1}{G}\sum_{g=1}^{G} w_{a,g}$, where $w_{a,g} = 1$ for a win and ties are broken by coins per the rules of the game. Chance level is $1/N$, so anything clearly above that means an agent is stronger than its opponents.

**Mean final score**: average Victory Points per game, $\bar{s}_a = \frac{1}{G}\sum_{g} s_{a,g}$. Because the real game is scored by going through VP for each category (science, military, cards, etc.) we will also break down mean final score in this way to see how each agent wins.

**Score margin**: points ahead of/behind the best opponent, $m_{a,g} = s_{a,g} - \max_{j \neq a} s_{j,g}$, averaged over games. This will be positive for an agent that wins often and negative for one that loses often. This should tell whether a game is a blowout or a close victory/defeat, and because it's less noisy we'll also use it as the objective for simulated annealing.

**Time per move**: average seconds per decision. This shows what the Monte Carlo agent's extra strength costs, so we'll run it at a few simulation budgets and compare it to the tuned heuristic agent, which should be close to instant during games.

The heuristic's weights and the Monte Carlo settings will be tuned on one set of seeds and evaluated on new ones, and we'll report 95% confidence intervals.

# Ethics & Privacy

Our project has few direct privacy concerns. Since all of our data will be coming from simulated games from our own AI agents, we will collect no information about real people. We will be not be using game logs from human players or from online platforms. If we do decide to add human gameplay data, we would only either use publicly available and released datasets with the terms that this would be for research use and player identifiers being removed, or use our own gameplay or the gameplay of those who consent to play and let us use the data.

An ethical risk we might face is misuse of the AI agent. With a strong 7 Wonders agent, the agent could be used to cheat in online games against human players which is unfair and would violate the terms of service of most online game platforms. To mitigate this risk on our end we will make it so that our code will only run inside our own simulated environment and that there is no interface to any online platform. Our repository and README will state that the project is intended for academic research purposes only.

We also have the responsibility to report our results that we get from our agents honestly. Our agents will only be tested and compared against each other and not against human players (if this changes, our human player data is subject to the restrictions we stated above) and our engine will simplify parts of the game such as automatically choosing the cheapest way to pay for a card. A high win rate or score in our simulated tournaments therefore does not mean an agent will play well against real people. We will state these limitations and test our engine's rules carefully.
